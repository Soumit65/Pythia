```python
import pythia8
import time
import matplotlib.pyplot as plt
import numpy as np
```

### 1. Hello world: one g g → t t̄ event
In this two gluons collide and turn into a top and an anti-top quark. The top is the eaviest fundamental particle and making it takes a lot of collision energy (Using 13 TeV in our simulation)

```python
pythia = pythia8.Pythia()
pythia.readString("Beams:eCM = 13000.")        # 13 TeV
pythia.readString("Top:gg2ttbar = on")         # switch on g g -> t tbar
pythia.readString("Next:numberCount = 0")      # no progress printing
pythia.readString("Print:quiet = on")          # shorter start-up output
pythia.readString("Next:numberShowEvent = 0")  # do not auto-print events
pythia.init()

ok = pythia.next()                              # generate one event; False if it aborted
print("PYTHIA version:", pythia.settings.parm("Pythia:versionNumber"))
print("event generated:", ok, "| entries in event record:", pythia.event.size())
```

### The event record

Each entry has a PDG particle id, a **status code** (why it was added) and a momentum.
Final-state particles have positive status. Particles that decayed or radiated get their status negated, but they stay in the record.
Below we print the first 12 entries and the total number of final-state particles.

### 2. Event loop: top pT and pseudorapidity

Generating many events. Finding the last top quark (id = 6) and recording its transverse momentum pT and pseudorapidity eta. We also add q q̄ → t t̄ so the sample is a mix of both production processes.

pseudorapidity: angle from the beam line 
$$ \eta = - \text{ln}\; \text{tan} \frac{\theta}{2}$$

$\theta$ is the angle from the beam




```python
N_EVENTS = 1000

pythia = pythia8.Pythia()
for s in ["Beams:eCM = 13000.", "Top:gg2ttbar = on", "Top:qqbar2ttbar = on",
          "Next:numberCount = 0", "Print:quiet = on", "Next:numberShowEvent = 0"]:
    pythia.readString(s)
pythia.init()

pt_top, eta_top = [], []
t0 = time.time()
for _ in range(N_EVENTS):
    if not pythia.next():
        continue
    iTop = 0
    for i in range(pythia.event.size()):
        if pythia.event[i].id() == 6:
            iTop = i                      # keeps the LAST top copy
    if iTop:
        pt_top.append(pythia.event[iTop].pT())
        eta_top.append(pythia.event[iTop].eta())
print(f"{len(pt_top)} events in {time.time()-t0:.1f} s")
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 3.8))
ax[0].hist(pt_top, bins=50, range=(0, 400), histtype="step", color="C0")
ax[0].set(xlabel="top pT (GeV)", ylabel="events", title="Top transverse momentum")
ax[1].hist(eta_top, bins=50, range=(-5, 5), histtype="step", color="C1")
ax[1].set(xlabel="top pseudorapidity", ylabel="events", title="Top pseudorapidity")
plt.tight_layout(); plt.show()
print(f"mean pT = {np.mean(pt_top):.1f} GeV")
```

Q2) Why is the eta distribution symmetric about zero and fairly wide? (Think about the two beams and the momentum fractions of the colliding partons.)

Ans) Both protons are identical, and each parton’s share of its proton’s energy is random. Gluon 1 is as likely to carry more as gluon 2. If gluon 1 is stronger, the top pair gets pushed toward proton 2’s direction, and the reverse is equally likely. No preferred direction means symmetric.

### 3. Charged multiplicity with FSR, ISR and MPI switched off one at a time

The charged multiplicity is the number of charged particles in the final state. Three ingredients add particles:

- **FSR**: radiation from the outgoing partons (`PartonLevel:FSR`)
- **ISR**: radiation from the incoming partons (`PartonLevel:ISR`)
- **MPI**: additional parton-parton collisions in the same pp event (`PartonLevel:MPI`)

We write a function that runs a fresh Pythia with a given list of extra settings and returns the multiplicity of each event, then compare four setups.


```python
def multiplicities(extra, n_events=1000):
    py = pythia8.Pythia()
    for s in ["Beams:eCM = 13000.", "Top:gg2ttbar = on", "Top:qqbar2ttbar = on",
              "Next:numberCount = 0", "Print:quiet = on", "Next:numberShowEvent = 0",
              "Random:setSeed = on", "Random:seed = 12345"] + extra:
        py.readString(s)
    py.init()
    out = []
    for _ in range(n_events):
        if not py.next():
            continue
        n = 0
        for i in range(py.event.size()):
            p = py.event[i]
            if p.isFinal() and p.isCharged():
                n += 1
        out.append(n)
    return np.array(out)

setups = {
    "all on":  [],
    "no FSR":  ["PartonLevel:FSR = off"],
    "no ISR":  ["PartonLevel:ISR = off"],
    "no MPI":  ["PartonLevel:MPI = off"],
}
N = 1000          # lower for a quick test, raise for smoother histograms
results = {}
for name, extra in setups.items():
    t0 = time.time()
    results[name] = multiplicities(extra, N)
    print(f"{name:8s} done in {time.time()-t0:5.1f} s")


    
```

```python
plt.figure(figsize=(7, 4.5))
for name, arr in results.items():
    plt.hist(arr, bins=60, range=(0, 600), histtype="step", label=f"{name} (mean {arr.mean():.1f})")
plt.xlabel("charged multiplicity"); plt.ylabel("events")
plt.title("pp -> t tbar at 13 TeV"); plt.legend(); plt.tight_layout(); plt.show()

print(f"{'setup':8s} {'mean':>8s} {'std':>8s} {'error on mean':>14s}")
for name, arr in results.items():
    print(f"{name:8s} {arr.mean():8.1f} {arr.std():8.1f} {arr.std()/np.sqrt(len(arr)):14.2f}")
```

Charged multiplicity” just means: how many electrically charged particles are in the final state.

Each radiated gluon or extra collision creates more coloured partons, and each ends up as a spray of hadrons. So switching off any of the three removes some particles.

MPI off: removes the whole extra layer of collisions, with lots of particles each. I’d predict the biggest drop.
ISR off or FSR off: each removes some radiated gluons. Each gluon adds a modest amount of particles. I’d predict smaller drops, likely similar to each other.

The multiplicity won’t go to zero in any case. The top and anti-top still decay into quarks that hadronize, and the leftover pieces of the protons still produce particles.
